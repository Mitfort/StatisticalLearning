# Zadanie 3 Test permutacyjny dla zmiennych niepowiązanych


```python
import numpy as np
```

```python
low_intensity = [378,346,245,285,365,245,208,360,296,224,292]
high_intensity = [218,264,211,180,256,240,261,205,145,195,187,210,378,204,232,237,310]

test_diff = abs(np.average(low_intensity) - np.average(high_intensity))

all = low_intensity + high_intensity
len(all)
```

```python
smaller_set_length = len(low_intensity) if len(low_intensity) <= len(high_intensity) else len(high_intensity)
avg_diffs = []

for _ in range(10000):
    all_copy = all.copy()
    first_list = []
    
    for i in range(smaller_set_length):
        random = np.random.choice(all_copy)
        first_list.append(random)
        all_copy.remove(random) 
    
    avg1 = np.average(first_list)
    avg2 = np.average(all_copy) 

    diff = abs(avg1 - avg2)
    avg_diffs.append(diff)

avg_diffs.sort()

lower_bound = avg_diffs[int(0.025 * len(avg_diffs))]
upper_bound = avg_diffs[int(0.975 * len(avg_diffs))]

print(f"Confidence Interval: ({lower_bound:.4f}, {upper_bound:.4f})")
```

```python
len(avg_diffs)
```

```python
counter = sum(1 for i in avg_diffs if i > test_diff)
higher_than = counter / len(avg_diffs)
print(f"Higher than base diff {test_diff} = {higher_than * 100}%")
```
