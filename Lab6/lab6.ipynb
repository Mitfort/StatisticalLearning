```python
rng = np.random.default_rng(seed=42)
X_raw = rng.normal(loc=0.0, scale=1.0, size=100)
epsilon = rng.normal(loc=0.0, scale=0.1, size=100)

Y = 2 + 3 * X_raw + 5 * X_raw**2 - 1.5 * X_raw**3 + epsilon

X = np.column_stack([X_raw,X_raw**2,X_raw**3])
```

```python
class ForwardStepwiseSelector:
    def __init__(self, max_features=3):
        """
            Execute the Forward Stepwise Selector\n
            :param max_features: Maximum number of features to select
        """
        self.max_features = max_features
        self.selected_features = []
        self.RSS_history = []

    def _RSS(self, y, yhat):
        """Calculate Residual Sum of Squares"""
        return np.sum((y - yhat) ** 2)
    
    def fit(self, X, y):
        n_samples, n_features = X.shape
        features = list(range(n_features))
        self.selected_features = []

        if self.max_features > n_features:
            self.max_features = n_features
        elif n_features == 0 or self.max_features == 0:
            return self

        for i in range(self.max_features):
            best_rss = np.inf
            best_feature = None

            for feature in features:
                if feature in self.selected_features:
                    continue

                current_features = self.selected_features + [feature]
                X_subset = X[:,current_features]

                coef = np.linalg.lstsq(X_subset, y, rcond=None)
                print(f"coef: {coef}")
                yhat = X_subset @ coef[0]

                rss = self._RSS(y, yhat)

                if rss < best_rss:
                    best_rss = rss
                    best_feature = feature

            self.selected_features.append(best_feature)
        return self
```

```python
model = ForwardStepwiseSelector(max_features=3)
model.fit(X,Y)

print(model.selected_features)
```
